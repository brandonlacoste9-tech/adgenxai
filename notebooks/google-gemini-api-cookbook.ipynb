### Integration with BeeHive Metrics Ritual

Every generation will include **metrics tracking** for continuous optimization:

```python
# Basic text generation with BeeHive Metrics Ritual tracking
import time
from dataclasses import dataclass
from typing import Dict, Any, Optional

@dataclass
class MetricsEvent:
    """Metrics Ritual: Track agent performance for BeeHive optimization"""
    agent_id: str
    model: str
    prompt_tokens: int
    completion_tokens: int
    latency_ms: float
    success: bool
    cost_estimate: float
    user_rating: Optional[float] = None
    session_id: str = ""
    
class GeminiAgentWithMetrics:
    """Gemini wrapper implementing BeeHive Metrics Ritual"""
    
    def __init__(self, model_name="gemini-2.5-flash", session_id=None):
        self.model_name = model_name
        self.model = genai.GenerativeModel(model_name)
        self.session_id = session_id or adgen_config.session_id
        self.metrics_history = []
        
    def generate_with_metrics(self, prompt: str, **kwargs) -> Dict[str, Any]:
        """Generate content with full metrics tracking"""
        
        start_time = time.time()
        agent_id = f"gemini-{self.model_name.replace('.', '-')}"
        
        try:
            # Generate content
            response = self.model.generate_content(prompt, **kwargs)
            
            # Calculate metrics
            end_time = time.time()
            latency_ms = (end_time - start_time) * 1000
            
            # Estimate token counts (simplified)
            prompt_tokens = len(prompt.split()) * 1.3  # Rough estimate
            completion_tokens = len(response.text.split()) * 1.3 if response.text else 0
            
            # Cost estimation (Gemini 2.5 Flash pricing)
            cost_per_1k_input = 0.000075  # $0.075 per 1M tokens
            cost_per_1k_output = 0.0003   # $0.30 per 1M tokens
            cost_estimate = (prompt_tokens * cost_per_1k_input / 1000) + \
                          (completion_tokens * cost_per_1k_output / 1000)
            
            # Create metrics event
            metrics = MetricsEvent(
                agent_id=agent_id,
                model=self.model_name,
                prompt_tokens=int(prompt_tokens),
                completion_tokens=int(completion_tokens),
                latency_ms=round(latency_ms, 2),
                success=True,
                cost_estimate=round(cost_estimate, 6),
                session_id=self.session_id
            )
            
            self.metrics_history.append(metrics)
            
            return {
                "response": response,
                "metrics": metrics,
                "text": response.text if response.text else "",
                "success": True
            }
            
        except Exception as e:
            # Track failed attempts
            end_time = time.time()
            latency_ms = (end_time - start_time) * 1000
            
            metrics = MetricsEvent(
                agent_id=agent_id,
                model=self.model_name,
                prompt_tokens=int(len(prompt.split()) * 1.3),
                completion_tokens=0,
                latency_ms=round(latency_ms, 2),
                success=False,
                cost_estimate=0.0,
                session_id=self.session_id
            )
            
            self.metrics_history.append(metrics)
            
            return {
                "response": None,
                "metrics": metrics,
                "text": "",
                "success": False,
                "error": str(e)
            }
    
    def get_performance_summary(self) -> Dict[str, Any]:
        """BeeHive Metrics: Get agent performance analytics"""
        if not self.metrics_history:
            return {"message": "No metrics data available"}
            
        total_requests = len(self.metrics_history)
        successful_requests = sum(1 for m in self.metrics_history if m.success)
        success_rate = (successful_requests / total_requests) * 100
        avg_latency = sum(m.latency_ms for m in self.metrics_history) / total_requests
        total_cost = sum(m.cost_estimate for m in self.metrics_history)
        
        return {
            "agent_id": self.metrics_history[0].agent_id,
            "model": self.model_name,
            "total_requests": total_requests,
            "success_rate": round(success_rate, 2),
            "avg_latency_ms": round(avg_latency, 2),
            "total_cost_usd": round(total_cost, 6),
            "session_id": self.session_id
        }

# Initialize agents for different models
flash_agent = GeminiAgentWithMetrics("gemini-2.5-flash")
pro_agent = GeminiAgentWithMetrics("gemini-2.5-pro")

print("🤖 Gemini agents initialized with BeeHive Metrics Ritual")
print("📊 All generations will be tracked for performance optimization")
print("🔬 Ready for Agent-First content generation!")
```
